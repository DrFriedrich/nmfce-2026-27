# Iterative linear solver

The Gauss-Seidel iteration to solve the linear algebraic equation system $A x = b$ is given by
$$
x_i^{(k+1)} = \frac{1}{a_{ii}}\left(b_i - \sum_{j=1}^{i-1} a_{ij}x_j^{(k+1)} - \sum_{j=i+1}^{N} a_{ij}x_j^{(k)}\right), \quad i = 1, \dots, N
$$

```python
import numpy as np

def Gauss_Seidel_step(x, A, b):
    """
    Function that returns the next step of the approximation of the solution x of Ax=b
    Inputs:
    x - Current solution
    A - Matrix with the coefficients of the linear system
    b - Vector with the right hand side
    """
    x_next = np.zeros_like(x, dtype=np.float64)
    for i in range(A.shape[0]):
        s1 = np.dot(A[i, :i], x_next[:i])
        s2 = np.dot(A[i, i + 1 :], x[i + 1 :])
        x_next[i] = (b[i] - s1 - s2) / A[i, i]
        
    return x_next
```

## Apply the Gauss-Seidel method to a diagonally dominant matrix

```python
from numpy.testing import assert_almost_equal

# Define the matrix, right hand side and starting guess
A = np.array([[6, -2, 3], [1, 4, -1], [-1, -3, 7]])
b = np.array([3, 15, -18])
x = np.array([1, 1, 1])

# Check one step of the Gauss-Seidel method
assert_almost_equal(Gauss_Seidel_step(x, A, b), [0.333, 3.917, -0.845], decimal=3)

# Print a header first
print(f"{'Iter':<7} | {'x_1':>8} | {'x_2':>8} | {'x_3':>8}")
print("-" * 40)

# Perform three steps of the Gauss-Seidel method
for i in range(6):
    x = Gauss_Seidel_step(x, A, b)
    print(f"Step {i:<2} | {x[0]:>8.4f} | {x[1]:>8.4f} | {x[2]:>8.4f}")
```

## Apply the Gauss-Seidel method to a matrix which is not diagonally dominant

```python
# Change the matrix and repeat
A = np.array([[1, -4, 2], [7, 3, -1], [-1, -3, 7]])
b = np.array([7, 5, 2])

# Print a header first
print(f"{'Iter':<7} | {'x_1':>10} | {'x_2':>10} | {'x_3':>10}")
print("-" * 46)

x = np.array([1, -1, 3])
for i in range(4):
    x = Gauss_Seidel_step(x, A, b)
    print(f"Step {i:<2} | {x[0]:>10.4f} | {x[1]:>10.4f} | {x[2]:>10.4f}")
```
